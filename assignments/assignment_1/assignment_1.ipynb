```python
# Download 'retina_blood_vessel_data.zip' directly using its Google Drive file ID.

import os
import subprocess

file_id = '1TTX4198s2Yx22LlSfg6dIvfrxuZHp3qa' # Extracted from the provided URL
file_name = 'retina_blood_vessel_data.zip'
output_path = os.path.join('/content', file_name)

print(f"Downloading {file_name} from Google Drive...")
download_command = f'gdown --id {file_id} -O "{output_path}"'
print(f"Executing: {download_command}")
result = subprocess.run(download_command, shell=True, capture_output=True, text=True)

if result.returncode == 0:
    print(f"Successfully downloaded {file_name} to {output_path}")
    print(result.stdout)
else:
    print(f"Error downloading {file_name}:")
    print(result.stderr)
```

```python
# Unzip the downloaded file into the expected directory and clean up.

import os
import zipfile

zip_file_path = os.path.join('/content', 'retina_blood_vessel_data.zip')
dest_path = '/content/unzipped_data'
if os.path.exists(zip_file_path):
    # Create the target directory if it doesn't exist
    os.makedirs(dest_path, exist_ok=True)

    # Unzip the file
    with zipfile.ZipFile(zip_file_path, 'r') as zip_ref:
        zip_ref.extractall(dest_path)
    print(f"Successfully unzipped '{zip_file_path}' to '{dest_path}'.")

    # Remove the zip file after extraction
    os.remove(zip_file_path)
    print(f"Removed '{zip_file_path}'.")
else:
    print(f"Error: '{zip_file_path}' not found. Download might have failed.")
```
